```python
# Step 3.5: Check the number of fields in every physical line

from collections import Counter

field_counts = Counter()
unusual_lines = []

with open(raw_file, "r", encoding="utf-8", errors="replace") as file:
    for line_number, line in enumerate(file, start=1):

        number_of_fields = len(line.rstrip("\r\n").split(","))

        field_counts[number_of_fields] += 1

        if number_of_fields != 10:
            unusual_lines.append(
                (line_number, number_of_fields, line.rstrip("\r\n"))
            )

print("Field-count distribution:")
print(dict(sorted(field_counts.items())))

print("\nTotal physical lines with a field count other than 10:")
print(len(unusual_lines))

print("\nFirst 20 unusual lines:")
for line_number, number_of_fields, content in unusual_lines[:20]:
    print(
        f"Line {line_number}: "
        f"{number_of_fields} fields | {content!r}"
    )
```

```python
# Step 3.6: Inspect records using Python's CSV parser

import csv
from collections import Counter

csv_field_counts = Counter()
csv_unusual_records = []

with open(
    raw_file,
    "r",
    encoding="utf-8",
    errors="replace",
    newline=""
) as file:

    reader = csv.reader(file)

    for record_number, row in enumerate(reader, start=1):

        csv_field_counts[len(row)] += 1

        if len(row) != 10:
            csv_unusual_records.append(
                (record_number, len(row), row)
            )

print("CSV parser field-count distribution:")
print(dict(sorted(csv_field_counts.items())))

print("\nTotal records with a field count other than 10:")
print(len(csv_unusual_records))

print("\nFirst 15 unusual records:")

for record_number, field_count, row in csv_unusual_records[:15]:
    print(f"\nRecord {record_number}: {field_count} fields")
    print(row)
```

```python
# Step 3.34: Identify extreme timing values

import numpy as np

# Select numeric timing features
timing_cols_fixed = [
    col for col in df_fixed.select_dtypes(include="number").columns
    if col not in ["session", "repetition", "total time"]
]

# Identify values with absolute magnitude greater than 100 seconds
extreme_mask = df_fixed[timing_cols_fixed].abs() > 100

# Find rows containing at least one extreme value
extreme_rows = df_fixed.loc[extreme_mask.any(axis=1)].copy()

print("Rows containing extreme timing values:", len(extreme_rows))

print("\nFirst 10 affected records:")
display(
    extreme_rows[
        ["participant", "session", "repetition"]
        + timing_cols_fixed
    ].head(10)
)

print("\nNumber of extreme values in each feature:")
print(
    extreme_mask.sum()
    .loc[lambda x: x > 0]
    .sort_values(ascending=False)
)
```

```python
# Step 3.35: Inspect exact extreme timing values

for idx, row in extreme_rows.iterrows():

    print("\n" + "=" * 60)
    print("DataFrame index:", idx)
    print("Participant:", row["participant"])
    print("Session:", row["session"])
    print("Repetition:", row["repetition"])

    print("\nExtreme timing values:")

    for col in timing_cols_fixed:
        value = row[col]

        if abs(value) > 100:
            print(f"{col}: {value}")
```

```python
# Step 3.37: Create a compact report of extreme timing values

extreme_report = []

for idx, row in extreme_rows.iterrows():

    for col in timing_cols_fixed:
        value = row[col]

        if abs(value) > 100:
            extreme_report.append({
                "dataframe_index": idx,
                "participant": row["participant"],
                "session": row["session"],
                "repetition": row["repetition"],
                "feature": col,
                "value": value
            })

extreme_report_df = pd.DataFrame(extreme_report)

display(extreme_report_df)
```

```python
# Step 4.00: Split data using separate sessions

# Training records come from session 1
train_mask = df_model["session"] == 1

# Testing records come from session 2
test_mask = df_model["session"] == 2

# Create training inputs and labels
X_train = X.loc[train_mask].copy()
y_train = y.loc[train_mask].copy()

# Create testing inputs and labels
X_test = X.loc[test_mask].copy()
y_test = y.loc[test_mask].copy()

print("Training data shape:", X_train.shape)
print("Training labels shape:", y_train.shape)

print("\nTesting data shape:", X_test.shape)
print("Testing labels shape:", y_test.shape)

print("\nParticipants in training:", y_train.nunique())
print("Participants in testing:", y_test.nunique())

print("\nTraining sessions:", df_model.loc[train_mask, "session"].unique())
print("Testing sessions:", df_model.loc[test_mask, "session"].unique())

# Check whether all testing participants appear in training
unseen_participants = set(y_test.unique()) - set(y_train.unique())

print("\nParticipants in testing but not training:", unseen_participants)
```

```python
# Step 4.01: Scale the timing features

from sklearn.preprocessing import StandardScaler

# Create the scaler
scaler = StandardScaler()

# Learn scaling parameters from training data only
X_train_scaled = scaler.fit_transform(X_train)

# Apply the same scaling parameters to testing data
X_test_scaled = scaler.transform(X_test)

print("Scaled training data shape:", X_train_scaled.shape)
print("Scaled testing data shape:", X_test_scaled.shape)

print("\nMean of first five scaled training features:")
print(X_train_scaled.mean(axis=0)[:5].round(3))

print("\nStandard deviation of first five scaled training features:")
print(X_train_scaled.std(axis=0)[:5].round(3))

print("\nAny NaN values in training data:", __import__("numpy").isnan(X_train_scaled).any())
print("Any NaN values in testing data:", __import__("numpy").isnan(X_test_scaled).any())
```

```python
# Step 4.02: Encode participant labels

from sklearn.preprocessing import LabelEncoder

# Create the label encoder
label_encoder = LabelEncoder()

# Learn participant labels from training data
y_train_encoded = label_encoder.fit_transform(y_train)

# Apply the same mapping to testing data
y_test_encoded = label_encoder.transform(y_test)

# Display the results
print("Training labels shape:", y_train_encoded.shape)
print("Testing labels shape:", y_test_encoded.shape)

print("\nTotal number of participants:", len(label_encoder.classes_))

print("\nFirst 10 participant IDs:")
print(label_encoder.classes_[:10])

print("\nFirst 10 encoded training labels:")
print(y_train_encoded[:10])

print("\nMinimum encoded label:", y_train_encoded.min())
print("Maximum encoded label:", y_train_encoded.max())
```

```python
# Step 4.03: Build and train the MLP model

from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score
import warnings
from sklearn.exceptions import ConvergenceWarning

# Create the MLP model
mlp_model = MLPClassifier(
    hidden_layer_sizes=(128, 64),
    activation="relu",
    solver="adam",
    batch_size=256,
    learning_rate_init=0.001,
    max_iter=50,
    early_stopping=False,
    random_state=42,
    verbose=True
)

print("Starting MLP model training...\n")

# Train the model using Session 1 data
with warnings.catch_warnings(record=True) as caught_warnings:
    warnings.simplefilter("always", ConvergenceWarning)
    mlp_model.fit(X_train_scaled, y_train_encoded)

    convergence_warnings = [
        warning for warning in caught_warnings
        if issubclass(warning.category, ConvergenceWarning)
    ]

print("\nModel training completed.")

print("Number of training iterations:", mlp_model.n_iter_)
print("Training data shape:", X_train_scaled.shape)
print("Number of input features:", mlp_model.n_features_in_)
print("Number of output classes:", len(mlp_model.classes_))

if convergence_warnings:
    print("\nNote: The model reached the iteration limit before convergence.")
    print("We will review the training results before deciding whether to adjust it.")
```

```python
# Step 4.04: Evaluate the MLP model

from sklearn.metrics import accuracy_score, balanced_accuracy_score

# Predict participant labels for Session 2
y_test_pred = mlp_model.predict(X_test_scaled)

# Calculate evaluation metrics
test_accuracy = accuracy_score(y_test_encoded, y_test_pred)
test_balanced_accuracy = balanced_accuracy_score(
    y_test_encoded,
    y_test_pred
)

# Calculate training accuracy for comparison
y_train_pred = mlp_model.predict(X_train_scaled)
train_accuracy = accuracy_score(y_train_encoded, y_train_pred)

# Display results
print("MODEL EVALUATION RESULTS")
print("-" * 35)

print(f"Training Accuracy: {train_accuracy * 100:.2f}%")
print(f"Testing Accuracy: {test_accuracy * 100:.2f}%")
print(f"Balanced Testing Accuracy: {test_balanced_accuracy * 100:.2f}%")

print("\nTesting samples:", len(y_test_encoded))
print("Correct predictions:", (y_test_encoded == y_test_pred).sum())
print("Incorrect predictions:", (y_test_encoded != y_test_pred).sum())
```
